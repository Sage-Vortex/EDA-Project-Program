```python
#This assignment is about joining dataframes, so does not apply on the mailed dataset
print("Assignment - 1")
import pandas as pd
import numpy as np

dataframe1 = pd.DataFrame({
    "ProductID": range(1, 11),
    "NameofProd": [
        "Pendrive", "CPU", "EduComp", "LED TV", "Electricity Generator",
        "Face-Scanner", "Xerox Machine", "Table", "Projector", "I-Pad"
    ],
    "Price": [520, 25000, 45000, 32000, 90000, 17000, 150000, 1500, 7200, 40000]
})

dataframe2 = pd.DataFrame({
    "ProductID": range(1, 11),
    "Company_Location": [
        "Nagasaki", "Nanking", "Athens", "Carthage", "Babylon",
        "Alexandria", "Waterloo", "Rome", "Pataliputra", "Corsica"
    ]
})

dataframe3 = pd.DataFrame({
    "ProductID": range(11, 21),
    "NameofProd": [
        "Telephone", "Apple Watch", "Marshalls Speakers", "Sandisk HardDrive",
        "Sandisk Aux Cable", "Wifi Extender", "Meta Glasses", "Type-C Charger",
        "Karaoke Set", "Extension Box"
    ],
    "Price": [8000, 25000, 10000, 900, 1100, 770, 13000, 500, 15000, 750]
})

print("Dataframe 1:")
print(dataframe1)

print("\nDataframe 2:")
print(dataframe2)

print("\nDataframe 3:")
print(dataframe3)

df1_3 = pd.concat([dataframe1, dataframe3], ignore_index=True)
print("\n1. Appending Dataframe 1 and Dataframe 3:")
print(df1_3)

df1_2 = pd.concat([dataframe1.reset_index(drop=True), dataframe2.drop(columns="ProductID").reset_index(drop=True)], axis=1)
print("\n2. Concatenating Dataframe 1 and Dataframe 2:")
print(df1_2)

merged_df1_2 = pd.merge(dataframe1, dataframe2, on="ProductID")
print("\n3. Merging Dataframe 1 and Dataframe 2:")
print(merged_df1_2)

inner_df = pd.merge(dataframe1, dataframe2, on="ProductID", how="inner")
print("\n4(a). Inner Join:")
print(inner_df)

left_df = pd.merge(dataframe1, dataframe2, on="ProductID", how="left")
print("\n4(b). Left Join:")
print(left_df)

right_df = pd.merge(dataframe1, dataframe2, on="ProductID", how="right")
print("\n4(c). Right Join:")
print(right_df)

fullOuter_df = pd.merge(dataframe1, dataframe2, on="ProductID", how="outer")
print("\n4(d). Full Outer Join:")
print(fullOuter_df)

Rajwardhan = np.array([203, 127, 285, 888, 725, 6, 491, 142, 775, 942])

print("\nOriginal Vector:")
print(Rajwardhan)

min_val = np.min(Rajwardhan)
max_val = np.max(Rajwardhan)

print("\nMinimum Value:", min_val)
print("Maximum Value:", max_val)

norm_3to3 = -3 + ((Rajwardhan - min_val) * (6 / (max_val - min_val)))
print("\n1(a). Normalization to [-3, +3]:")
print(norm_3to3)

norm_1to1 = -1 + ((Rajwardhan - min_val) * (2 / (max_val - min_val)))
print("\n1(b). Normalization to [-1, +1]:")
print(norm_1to1)

norm_0to1 = (Rajwardhan - min_val) / (max_val - min_val)
print("\n1(c). Min-Max Normalization to [0, 1]:")
print(norm_0to1)

z_score = (Rajwardhan - np.mean(Rajwardhan)) / np.std(Rajwardhan, ddof=1)
print("\n2. Z-Score Normalization:")
print(z_score)

j = len(str(np.max(np.abs(Rajwardhan))))
decimal_scaling = Rajwardhan / (10 ** j)
print("\n3. Decimal Scaling Normalization:")
print(decimal_scaling)
```

```python
#This assignment was about Time Series and since the mailed data is not applicable, another data has been used 
print("Assignment - 6")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

air_passengers = [
    112,118,132,129,121,135,148,148,136,119,104,118,
    115,126,141,135,125,149,170,170,158,133,114,140,
    145,150,178,163,172,178,199,199,184,162,146,166,
    171,180,193,181,183,218,230,242,209,191,172,194,
    196,196,236,235,229,243,264,272,237,211,180,201,
    204,188,235,227,234,264,302,293,259,229,203,229,
    242,233,267,269,270,315,364,347,312,274,237,278,
    284,277,317,313,318,374,413,405,355,306,271,306,
    315,301,356,348,355,422,465,467,404,347,305,336,
    340,318,362,348,363,435,491,505,404,359,310,337,
    360,342,406,396,420,472,548,559,463,407,362,405,
    417,391,419,461,472,535,622,606,508,461,390,432
]

dates = pd.date_range(start="1949-01-01", periods=144, freq="MS")

air_df = pd.DataFrame({
    "Date": dates,
    "Passengers": air_passengers
}).set_index("Date")

print("\n1. AirPassengers Time-Series Data")
print(air_df.head())

print("\n2. First Five Observations")
print(air_df.head())

print("\nLast Five Observations")
print(air_df.tail())

print("\n3. Start Time:", air_df.index.min())
print("End Time:", air_df.index.max())

print("\n4. Frequency: 12 observations per year")
print("Number of Observations:", len(air_df))
print("\nTime-Series Data:")
print(air_df)

plt.figure(figsize=(12, 5))
plt.plot(air_df.index, air_df["Passengers"])
plt.title("BDS0203 Monthly Air Passengers")
plt.xlabel("BDS0203 Year")
plt.ylabel("BDS0203 Number of Passengers")
plt.show()

values = air_df["Passengers"].astype(float)
series = values.to_numpy()

trend = values.rolling(window=12, center=True).mean()
trend = trend.rolling(window=2, center=True).mean()

detrended = values - trend
seasonal_by_month = detrended.groupby(detrended.index.month).mean()
seasonal_by_month = seasonal_by_month - seasonal_by_month.mean()

seasonal = pd.Series(
    [seasonal_by_month[m] for m in values.index.month],
    index=values.index
)

residual = values - trend - seasonal

decomposed_data = pd.DataFrame({
    "Observed": values,
    "Trend": trend,
    "Seasonal": seasonal,
    "Random": residual
})

print("\n5. Time-Series Decomposition")
print(decomposed_data)

fig, axes = plt.subplots(4, 1, figsize=(12, 10), sharex=True)

axes[0].plot(values)
axes[0].set_title("Observed")

axes[1].plot(trend)
axes[1].set_title("Trend")

axes[2].plot(seasonal)
axes[2].set_title("Seasonal")

axes[3].plot(residual)
axes[3].set_title("Random / Residual")

plt.tight_layout()
plt.show()

def calculate_acf(data, max_lag):
    data = np.asarray(data, dtype=float)
    data = data - np.mean(data)
    denominator = np.sum(data ** 2)
    result = [1.0]

    for lag in range(1, max_lag + 1):
        result.append(
            np.sum(data[lag:] * data[:-lag]) / denominator
        )

    return np.array(result)

def calculate_pacf(data, max_lag):
    data = np.asarray(data, dtype=float)
    acf_values = calculate_acf(data, max_lag)
    pacf_values = [1.0]

    for lag in range(1, max_lag + 1):
        matrix = np.empty((lag, lag))

        for i in range(lag):
            for j in range(lag):
                matrix[i, j] = acf_values[abs(i - j)]

        target = acf_values[1:lag + 1]
        coefficients = np.linalg.lstsq(
            matrix, target, rcond=None
        )[0]

        pacf_values.append(coefficients[-1])

    return np.array(pacf_values)

max_lag = 24
acf_values = calculate_acf(series, max_lag)
pacf_values = calculate_pacf(series, max_lag)
confidence_limit = 1.96 / np.sqrt(len(series))
lags = np.arange(max_lag + 1)

print("\n6. Autocorrelation Values:")
print(pd.Series(acf_values, index=lags))

plt.figure(figsize=(10, 5))
plt.stem(lags, acf_values)
plt.axhline(confidence_limit, linestyle="--", color="red")
plt.axhline(-confidence_limit, linestyle="--", color="red")
plt.title("BDS0203 Autocorrelation of AirPassengers")
plt.xlabel("Lag")
plt.ylabel("ACF")
plt.show()

print("\n7. Partial Autocorrelation Values:")
print(pd.Series(pacf_values, index=lags))

plt.figure(figsize=(10, 5))
plt.stem(lags, pacf_values)
plt.axhline(confidence_limit, linestyle="--", color="red")
plt.axhline(-confidence_limit, linestyle="--", color="red")
plt.title("BDS0203 Partial Autocorrelation of AirPassengers")
plt.xlabel("Lag")
plt.ylabel("PACF")
plt.show()

differences = np.diff(series)

def arima_111_residuals(params, data):
    intercept, phi, theta = params
    errors = np.zeros(len(data))
    residuals = np.zeros(len(data))

    for i in range(1, len(data)):
        residuals[i] = (
            data[i] - intercept
            - phi * data[i - 1]
            - theta * errors[i - 1]
        )
        errors[i] = residuals[i]

    return residuals

def arima_111_loss(params, data):
    residuals = arima_111_residuals(params, data)
    return np.mean(residuals[1:] ** 2)

params = np.array([np.mean(differences), 0.0, 0.0])
best_loss = arima_111_loss(params, differences)
steps = np.array([1.0, 0.2, 0.2])

for _ in range(150):
    improved = False

    for j in range(3):
        for direction in [-1, 1]:
            candidate = params.copy()
            candidate[j] += direction * steps[j]
            candidate[1] = np.clip(candidate[1], -0.98, 0.98)
            candidate[2] = np.clip(candidate[2], -0.98, 0.98)

            loss = arima_111_loss(candidate, differences)

            if loss < best_loss:
                params = candidate
                best_loss = loss
                improved = True

    if not improved:
        steps *= 0.7

print("\n8. Approximate ARIMA(1,1,1) Model")
print("Estimated Intercept:", params[0])
print("Estimated AR(1) Coefficient:", params[1])
print("Estimated MA(1) Coefficient:", params[2])
print("Residual Mean Squared Error:", best_loss)

forecast_differences = []
errors = arima_111_residuals(params, differences)
last_difference = differences[-1]
last_error = errors[-1]

for _ in range(12):
    next_difference = (
        params[0]
        + params[1] * last_difference
        + params[2] * last_error
    )

    forecast_differences.append(next_difference)
    last_difference = next_difference
    last_error = 0.0

forecast_values = series[-1] + np.cumsum(forecast_differences)

residual_variance = np.var(
    arima_111_residuals(params, differences)[1:],
    ddof=1
)

forecast_se = np.sqrt(
    residual_variance * np.arange(1, 13)
)

forecast_dates = pd.date_range(
    start="1961-01-01",
    periods=12,
    freq="MS"
)

forecast_df = pd.DataFrame({
    "Forecast": forecast_values,
    "Standard_Error": forecast_se,
    "Lower_95%": forecast_values - 1.96 * forecast_se,
    "Upper_95%": forecast_values + 1.96 * forecast_se
}, index=forecast_dates)

print("\n9. Forecast for 1961:")
print(forecast_df)

plt.figure(figsize=(12, 6))
plt.plot(
    air_df.index,
    air_df["Passengers"],
    label="Actual",
    color="darkgreen"
)
plt.plot(
    forecast_df.index,
    forecast_df["Forecast"],
    linestyle="--",
    label="Forecast",
    color="blue"
)
plt.plot(
    forecast_df.index,
    forecast_df["Upper_95%"],
    linestyle=":",
    color="gray",
    label="95% Confidence Interval"
)
plt.plot(
    forecast_df.index,
    forecast_df["Lower_95%"],
    linestyle=":",
    color="gray"
)
plt.fill_between(
    forecast_df.index,
    forecast_df["Lower_95%"],
    forecast_df["Upper_95%"],
    alpha=0.15
)
plt.title("BDS0203 AirPassengers: Actual and Forecasted Values")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.xlim(pd.Timestamp("1949-01-01"), pd.Timestamp("1962-01-01"))
plt.legend()
plt.show()
```

```python
print("Assignment - 8")
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import pdist

print(auto.head())
print(auto.isnull().sum())

target = auto.iloc[:, -1]
print(target.value_counts())

data = auto.dropna().copy()

numeric_data = data.select_dtypes(include=np.number)
numeric_data = numeric_data.loc[:, numeric_data.nunique() > 1]

print(numeric_data.head())

scaler = StandardScaler()
scaled_data = scaler.fit_transform(numeric_data)
scaled_df = pd.DataFrame(scaled_data, columns=numeric_data.columns, index=numeric_data.index)

print(scaled_df.head())

kmeans2 = KMeans(n_clusters=2, n_init=25, random_state=123)
kmeans2.fit(scaled_df)

print("K-Means with 2 clusters")
print("Cluster sizes:", pd.Series(kmeans2.labels_ + 1).value_counts().sort_index().values)
print("Cluster centers:")
print(pd.DataFrame(kmeans2.cluster_centers_, columns=numeric_data.columns))
print("Within-cluster sum of squares:", kmeans2.inertia_)
print("Percentage of variance explained:", round((1 - kmeans2.inertia_ / np.sum((scaled_data - scaled_data.mean(axis=0)) ** 2)) * 100, 2), "%")

wcss = []
for i in range(1, 11):
    model = KMeans(n_clusters=i, n_init=25, random_state=123)
    model.fit(scaled_df)
    wcss.append(model.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Plot")
plt.xticks(range(1, 11))
plt.grid(True, alpha=0.3)
plt.show()

final_kmeans = KMeans(n_clusters=3, n_init=25, random_state=123)
final_kmeans.fit(scaled_df)

print("K-Means with 3 clusters")
print("Cluster sizes:")
print(pd.Series(final_kmeans.labels_ + 1).value_counts().sort_index())
print("Cluster centers:")
print(pd.DataFrame(final_kmeans.cluster_centers_, columns=numeric_data.columns))
print("Within-cluster sum of squares:", final_kmeans.inertia_)
print("Percentage of variance explained:", round((1 - final_kmeans.inertia_ / np.sum((scaled_data - scaled_data.mean(axis=0)) ** 2)) * 100, 2), "%")

auto_kmeans = data.copy()
auto_kmeans["KMeans_Cluster"] = final_kmeans.labels_ + 1
print(auto_kmeans["KMeans_Cluster"].value_counts().sort_index())

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=auto_kmeans,
    x="Mileage",
    y="weight",
    hue="KMeans_Cluster",
    palette="Set1",
    s=60
)
plt.title("K-Means Clusters: Mileage vs Weight")
plt.show()

numeric_data_hc = data.select_dtypes(include=np.number)
numeric_data_hc = numeric_data_hc.loc[:, numeric_data_hc.nunique() > 1]

scaled_hc = StandardScaler().fit_transform(numeric_data_hc)

distance = pdist(scaled_hc, metric="euclidean")
print("First 100 pairwise Euclidean distances:")
print(distance[:100])

hc = linkage(scaled_hc, method="ward", metric="euclidean")
print("Hierarchical clustering completed.")
print("Number of observations:", len(data))

plt.figure(figsize=(16, 7))
dendrogram(hc, no_labels=True)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Observations")
plt.ylabel("Height")
plt.show()

hierarchical_clusters = fcluster(hc, t=3, criterion="maxclust")

print("Hierarchical cluster assignments:")
print(hierarchical_clusters)
print("Cluster sizes:")
print(pd.Series(hierarchical_clusters).value_counts().sort_index())

auto_hc = data.copy()
auto_hc["Hierarchical_Cluster"] = hierarchical_clusters

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=auto_hc,
    x="Mileage",
    y="weight",
    hue="Hierarchical_Cluster",
    palette="Set1",
    s=60
)
plt.title("Hierarchical Clusters: Mileage vs Weight")
plt.show()
```

```python
print("Assignment - 9")
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

pca_columns = ["Mileage", "displacement", "horsepower", "PowerWeightRatio", "weight", "acceleration"]
cars_pca = auto[pca_columns].dropna().copy()

print(cars_pca.describe())

scaler = StandardScaler()
cars_scaled = scaler.fit_transform(cars_pca)

pca_result = PCA()
pca_scores = pca_result.fit_transform(cars_scaled)

print("Standard deviations:")
print(np.sqrt(pca_result.explained_variance_))

print("Rotation / Principal Component Loadings:")
rotation = pd.DataFrame(
    pca_result.components_.T,
    index=pca_columns,
    columns=[f"PC{i+1}" for i in range(len(pca_columns))]
)
print(rotation)

eigenvalues = pca_result.explained_variance_
variance_percent = pca_result.explained_variance_ratio_ * 100
cumulative_variance = np.cumsum(variance_percent)

variance_table = pd.DataFrame({
    "PC": [f"PC{i+1}" for i in range(len(eigenvalues))],
    "Eigenvalue": eigenvalues,
    "Variance_Percent": variance_percent,
    "Cumulative_Percent": cumulative_variance
})

print(variance_table)

num_components = np.where(cumulative_variance >= 80)[0][0] + 1
print("Number of components required to explain at least 80% variance:", num_components)

plt.figure(figsize=(8, 5))
plt.plot(
    range(1, len(eigenvalues) + 1),
    eigenvalues,
    marker="o",
    linestyle="-"
)
plt.title("Scree Plot")
plt.xlabel("Principal Component")
plt.ylabel("Eigenvalue")
plt.xticks(range(1, len(eigenvalues) + 1))
plt.grid(True, alpha=0.3)
plt.show()

print("All principal component loadings:")
print(rotation.round(3))

print("First two principal component loadings:")
print(rotation.iloc[:, :2].round(3))

pca_plot = pd.DataFrame(
    pca_scores[:, :2],
    columns=["PC1", "PC2"],
    index=cars_pca.index
)

plt.figure(figsize=(9, 6))
sns.scatterplot(data=pca_plot, x="PC1", y="PC2", s=60)
plt.title("PCA Scores Plot")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.axhline(0, color="gray", linewidth=0.7)
plt.axvline(0, color="gray", linewidth=0.7)
plt.show()

plt.figure(figsize=(10, 7))
for i, feature in enumerate(pca_columns):
    plt.arrow(
        0, 0,
        rotation.iloc[i, 0],
        rotation.iloc[i, 1],
        head_width=0.03,
        length_includes_head=True
    )
    plt.text(
        rotation.iloc[i, 0] * 1.12,
        rotation.iloc[i, 1] * 1.12,
        feature,
        fontsize=9
    )

plt.xlim(-1, 1)
plt.ylim(-1, 1)
plt.axhline(0, color="gray", linewidth=0.7)
plt.axvline(0, color="gray", linewidth=0.7)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Biplot")
plt.grid(True, alpha=0.3)
plt.show()
```
